```python
import pandas as pd
import os
import pathlib

# This script concatenates CSV files from different directories into a single CSV file for each attribute.
# It reads CSV files from a source directory, processes them, and saves the concatenated data to a destination directory.
```

```python
# Root directory for the project
root = os.getcwd()
# Source directory containing CSV files
src_root = os.path.join(root, 'csv')
src_root = pathlib.Path(src_root)
assert os.path.exists(src_root)
# Destination directory for concatenated CSV files
dst_dir = os.path.join(root, 'timeseries_by_attribute')
if not os.path.exists(dst_dir):
    os.makedirs(dst_dir)

# Listing all directories in the source directory, except for ipynb checkpoints
dirs = [item for item in src_root.iterdir() if not 'ipynb' in item.stem]
```

```python
# Process each directory
for directory in dirs:
    stem = directory.stem
    if stem != 'pet':
        continue
        
    files = [item for item in directory.iterdir() if not 'ipynb' in item.stem]

    dfs = []

    for file in files:
        df = pd.read_csv(file)
        df['date'] = pd.to_datetime(df['date']) 
        df = df.set_index('date')
        dfs.append(df)
        
    result = pd.concat(dfs)
    result = result.sort_index()
    result = result.loc["1961":"2023"]
    # Drop the hours from the index
    result.index = result.index.date
    result.index =  result.index.set_names('date')
    dst_path = os.path.join(dst_dir, stem + '.csv')
    result.to_csv(dst_path)

# Display the concatenated data
result
```
