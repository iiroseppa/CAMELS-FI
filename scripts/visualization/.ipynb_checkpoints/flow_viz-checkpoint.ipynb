```python
import pandas as pd
#import geopandas as geopd
import os

import matplotlib.pyplot as plt
plt.rcParams["figure.dpi"]  = 200
plt.style.use('dark_background')
```

```python
#test_flows = pd.read_parquet("flow/flows_from_1995.parquet")
test_flows = pd.read_parquet("flow/all_flows.parquet")

# matlotlib uses inches
cm_conversion = 1 / 2.54 
```

```python
# Xlim units is days from 1.1.1970
# inarjoki, karigasniemi = 1409
# Aurajoki = 1117
gauge = '1409'

# light
color = '#4ca9ff'
#dark
#color = '#11548a'

ax = test_flows[[gauge]].plot.line(figsize=(12, 4), color=color, xlim=("1961", "2020"), alpha=1, linewidth=0.6, legend=False)
#plt.axis('off')
#path = os.path.join("/home/iiro/Documents/DIWA/visualizations/presentations/kuvat/flow_graph/", f"{gauge}.png")
#plt.savefig(path, transparent=True)
```

```python
ax = diff.plot.line(figsize=(12, 4), color=color, xlim=('1988', '1991'), alpha=1, linewidth=1, legend=False)
ax.grid()
```
