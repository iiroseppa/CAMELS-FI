```python
import pandas as pd
import os

import matplotlib.pyplot as plt

plt.rcParams["figure.dpi"]  = 400
plt.style.use('dark_background')
```

```python
hydro_path = '/home/iielse/Documents/DIWA/data/backups/CAMELS-FI_6_6_25/data/CAMELS_FI_hydrologic_attributes.csv'
hydro = pd.read_csv(hydro_path)
hydro = hydro.set_index('gauge_id')
clim_path = '/home/iielse/Documents/DIWA/data/backups/CAMELS-FI_6_6_25/data/CAMELS_FI_climatic_attributes.csv'
clim = pd.read_csv(clim_path)
clim = clim.set_index('gauge_id')

meta_path = '/home/iielse/Documents/DIWA/data/backups/CAMELS-FI_6_6_25/data/CAMELS_FI_meta_attributes.csv'
meta = pd.read_csv(meta_path)
meta = meta.set_index('gauge_id')
```

```python
hydro
```

```python
clim
```

```python
# Budyko curve is created by plotting Aridity (PE / P) compared to Runoff coefficient (Q/P)
hydroclim = clim.join([hydro, meta[['water_region_name', 'water_region_code']]])
hydroclim
bud = hydroclim[['p_mean', 'pet_mean', 'q_mean', 'water_region_name', 'water_region_code']].copy()
bud['aridity'] = bud.pet_mean / bud.p_mean
bud['runoff_coef'] = bud.q_mean / bud.p_mean
bud['rgb'] = bud['water_region_code'].map({2:'#4ca9ff', 1:'#11548a', 3:'#ffffff'})
bud
```

```python
lines = [
    ((0, 1), (1, 0)), 
    ((0, 1), (1.6, 1))
]
ax = bud.plot(x='aridity', y='runoff_coef', s=2, c='rgb', kind='scatter')
for (x1, y1), (x2, y2) in lines:
    ax = plt.plot([x1, x2], [y1, y2], ':', c='#aaaaaa')
```
