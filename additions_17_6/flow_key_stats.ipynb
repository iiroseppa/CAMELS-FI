```python
import pandas as pd
import geopandas as geopd
#import geopandas as geopd

import matplotlib.pyplot as plt
plt.rcParams["figure.dpi"]  = 400
plt.style.use('dark_background')
```

```python
# This script calculates key statistics for flow data.
# It reads flow data from a Parquet file, filters it to a specific time period,
# and calculates the number of valid observations for each gauge.

# Read the flow data for 1994-2016
flows = pd.read_parquet("/path/to/flows_from_1994.parquet")
flows = flows['1994-01-01':'2016-12-31']

places = geopd.read_file("/path/to/measurement_locations.gpkg", layer="test_stations")

place_ids = list(places['Paikka_Id'])

# Filter the flow data to include only the test stations
flows = flows[place_ids]
flows #display
```

```python
# Calculate the total number of valid observations
flows.count().sum()
```

```python
# Plot the measurement locations
places.plot()
```
