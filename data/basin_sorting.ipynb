```python
import geopandas as geopd
import pandas as pd

import matplotlib.pyplot as plt
plt.rcParams['figure.dpi'] = 400
plt.style.use('dark_background')

# This script processes and sorts basin data based on their areas.
# It reads basin data from a GeoPackage, calculates the area, sorts the basins by area, and saves the sorted data back to a GeoPackage.
```

```python
path = "/path/to/suitable_watersheds.gpkg"
basins = geopd.read_file(path, layer='all')
```

```python
basins.plot()
```

```python
basins.area.plot()
```

```python
basins['area_m2'] = basins.area
```

```python
# Sort the basins by area in descending order and reset the index
sorted_basins = basins.sort_values('area_m2', ascending=False, ignore_index=True)
```

```python
sorted_basins.head(10)
```

```python
# Save the sorted basins to a new GeoPackage
sorted_basins.to_file("/path/to/CAMELS-FI_basins.gpkg", driver='GPKG', layer='v1')
```
